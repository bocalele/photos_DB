```python
from geopy.geocoders import Nominatim
from geopy.point import Point

loc = Nominatim(user_agent="GetLoc")

# En partant du nom de la commune
getLoc = loc.geocode("Marseille")
print (getLoc.latitude, '  ', getLoc.longitude)
locname = loc.reverse(str(getLoc.latitude)+","+str (getLoc.longitude))
print (locname.raw['address'])

p = Point (getLoc.latitude, getLoc.longitude,0)
str(p)
# printing latitude and longitude
##print("Latitude = ", getLoc.latitude, "\n")
#print("Longitude = ", getLoc.longitude)
```
